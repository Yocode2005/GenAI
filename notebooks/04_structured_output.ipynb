## Solution is : Structure Output

```python
from pydantic import BaseModel, Field # pydantic is python library for data validation and In GenAI, especially when you're using LangChain, we use Pydantic mainly to make the LLM's output structured, validated, and predictable.
from typing import List

class ResponseStructure(BaseModel):
    name: str = Field(description="The name of the person")
    email: str = Field(description="The email address of the person")
    age: int = Field(description="The age of the person")

structured_llm = llm.with_structured_output(ResponseStructure) # hmne llm ko bta hai ki ab output response structure ke according aayega
```

```python
res = structured_llm.invoke(f"please give me only name, email and age from this text: {text}")
```

```python
class Movies(BaseModel):
    title: str = Field(description="The title of the movie")
    year: int = Field(description="The release year of the movie")

class MovieList(BaseModel):
    movies: List[Movies] = Field(description="A list of movies")

movie_llm = llm.with_structured_output(Movies)
movieList_llm = llm.with_structured_output(MovieList)
```
