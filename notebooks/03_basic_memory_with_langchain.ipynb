```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
)
```

```python
res = llm.invoke("Hello, My name is yogesh kumar meena.")
res.content
```

```python
res = llm.invoke("what is my name?")
res.content[0]["text"]
```

# memory concept

```python
#llms do not have memory, so it will not remember the previous conversation. To add memory.

prompts = [
    {"role": "user","content": "Hello, My name is yogesh kumar meena."},
    {"role": "assistant","content": "Hello, yogesh kumar meena! How can I assist you today?"},
    {"role": "user","content": "what is my name?"}
]
res = llm.invoke(prompts)
res.content[0]["text"]
```

### Manage History

```python
history = []
while True:
    query = input("User: ")
    if query.lower() in ["exit", "quit", "bye"]:
        print("Goodbye👋")
        break
    history.append({"role": "user", "content": query})
    print("User:", query)

    res = llm.invoke(history)

    history.append({"role": "assistant", "content": res.content[0]["text"]})
    print("Assistant:", res.content[0]["text"])
```
