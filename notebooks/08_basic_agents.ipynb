```python
from langchain_core.tools import tool
```

### How to define Tool

```python
@tool # python functions ko tool bnane ke liye
def add_numbers(a:int,b:int):# docstring triple commans me assign hoti hai ye bs btati hai ki function ya tool kya krne wala hai
    """ 
        it will return the sum of two numbers

        Args:
            a : Number One
            b : NUmber Two
    """
    return a+b

#tool-2
@tool 
def multiply_numbers(a:int,b:int):
    """ 
        it will return the product of two numbers

        Args:
            a : Number One
            b : NUmber Two
    """
    return a*b
```

```python
add_numbers.invoke({"a":12,"b":3})
```

### How to connect Tool with LLMs to make Agents

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
)
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="you are a math teacher, and always use tool for calculation."
)
```

```python
response = agent.invoke({"messages":[{"role":"user", "content":"What is 2 + 3 ?"}]})
```
