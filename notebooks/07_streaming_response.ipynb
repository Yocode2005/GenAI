```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    streaming=True
)
```

```python
question = "Distance  of earth from sun?"
```

```python
res = llm.stream(question) # stream ans genrate nhi krega ye sirf generator provide krvata hai mtlb jese jese response generate hota jayga vese vese respose hmme deta jay ga 
for chunk in res: # genrator gives chunk of response
    print(chunk.content,end="")
```
