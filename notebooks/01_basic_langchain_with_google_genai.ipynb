### static prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.7-flash",
)
```

```python
prompts = [ ### it is list of toples()()
    ("system","you are a python developer"), ### first argument me dete hai (type) mtlb kiska prompt hai ( system, assistant,user)
    ("user","How to sort the array ? expalin only in 3 lines"), ### second argument me dete hai (content) mtlb kya prompt hai
    ]
res  = llm.invoke(prompts)
for item in res.content:
    print("Type:", item["type"])
    print("Answer:", item["text"])
```
