### Dynamic prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import strOutputParser
# strOutputParser only gives the output in string format
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
)
out = strOutputParser()
```

```python
# fro upper case the output
def transform_case(result : str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
   {"role" : "system","content" : "You are a tranlator that translates {language1} to {language2}."},
   {"role" : "user","content" : "{query}"},
])
```

```python
chains = prompts | llm  | out | transform_case
# langchain me chainging ese hota hai ki prompts ka output llm ke input me chla jayga or llm ka output out ke input me chla jayga
# prompts ka output llm ke input me chla jayga or llm ka output out ke input me chla jayga
```
